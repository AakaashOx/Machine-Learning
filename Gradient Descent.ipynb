```python
import numpy as np
import matplotlib.pyplot as plt

# Simple dataset (you can replace with your own)
X = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 6, 8, 10])   # y = 2x

# Initialize parameters
m = 0   # slope
b = 0   # intercept

learning_rate = 0.01
epochs = 100

n = len(X)

# -------- Gradient Descent --------
for i in range(epochs):
    y_pred = m * X + b
    
    # Calculate gradients
    dm = (-2/n) * sum(X * (y - y_pred))
    db = (-2/n) * sum(y - y_pred)
    
    # Update parameters
    m = m - learning_rate * dm
    b = b - learning_rate * db

# Final values
print("Slope (m):", m)
print("Intercept (b):", b)
```

```python
# -------- Plot --------
plt.scatter(X, y)
plt.plot(X, m * X + b)
plt.title("Gradient Descent Regression")
plt.show()
```
